# 📖 Diccionario de Métodos y Palabras Reservadas

Aquí tienes una explicación sencilla de los conceptos y términos especiales de Python que hemos utilizado en este notebook:

### 1. `deque` (de la biblioteca `collections`)
* **¿Qué es?:** Es un tipo de lista optimizada (Double-Ended Queue o Cola Doblemente Terminada).
* **¿Para qué sirve?:** A diferencia de las listas normales, `deque` permite agregar y quitar elementos de forma extremadamente rápida tanto por el principio como por el final. Es ideal para implementar **Colas (FIFO)**.

### 2. `popleft()`
* **¿Qué es?:** Un método exclusivo de los objetos `deque`.
* **¿Para qué sirve?:** Elimina y devuelve el **primer** elemento de la cola (el de la izquierda). Es el comando clave para la acción de "Atender Cliente" en nuestro sistema.

### 3. `global`
* **¿Qué es?:** Una palabra reservada de Python.
* **¿Para qué sirve?:** Le indica a una función que la variable que va a modificar (en este caso, `turno`) no es una variable local nueva, sino que pertenece al entorno general (ámbito global) del programa.

### 4. `enumerate(lista, start=1)`
* **¿Qué es?:** Una función integrada de Python.
* **¿Para qué sirve?:** Nos permite recorrer una lista en un ciclo `for` obteniendo al mismo tiempo el elemento y su número de posición (índice), comenzando a contar desde el número que le indiquemos en `start` (en nuestro caso, `1` en lugar de `0`).

### 5. `try` y `except ValueError`
* **¿Qué es?:** Bloques para el manejo de excepciones (errores).
* **¿Para qué sirve?:** Evitan que el programa se detenga bruscamente (se "caiga") si el usuario ingresa texto cuando el sistema esperaba un número. Si ocurre un error de valor (`ValueError`), el bloque `except` lo captura y muestra un mensaje amigable.

### 6. `pop(indice)`
* **¿Qué es?:** Un método de las listas de Python.
* **¿Para qué sirve?:** Elimina el elemento de la lista que se encuentra en la posición especificada por el `indice` y lo devuelve. Lo usamos para quitar una tarea específica de nuestra lista.
